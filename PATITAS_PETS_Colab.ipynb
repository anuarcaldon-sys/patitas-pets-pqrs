## Paso 1 · Opción B — Subir el archivo `.zip`
Al ejecutar la celda aparece el botón **Elegir archivos**: seleccione `PATITAS-PETS_Entrega_Final_v3.zip`.

```python
%cd /content
from google.colab import files
import zipfile, shutil, os, glob

subidos = files.upload()
nombre_zip = next(n for n in subidos if n.lower().endswith(".zip"))

# Se extrae en una carpeta temporal y se ubica la raíz del proyecto (la que contiene src/main.py)
TEMP = "/content/_zip_patitas"
shutil.rmtree(TEMP, ignore_errors=True)
with zipfile.ZipFile(nombre_zip) as z:
    z.extractall(TEMP)
raiz_zip = os.path.dirname(os.path.dirname(glob.glob(f"{TEMP}/**/src/main.py", recursive=True)[0]))

shutil.rmtree("/content/patitas-pets-pqrs", ignore_errors=True)
shutil.move(raiz_zip, "/content/patitas-pets-pqrs")
shutil.rmtree(TEMP, ignore_errors=True)
print("Archivo cargado:", nombre_zip)
print("Contenido extraído:", sorted(os.listdir("/content/patitas-pets-pqrs")))
```

## Paso 3 — Ejecutar el programa

```python
import os, sys
from IPython.display import clear_output

RAIZ = "/content/patitas-pets-pqrs"
os.chdir(RAIZ)
if f"{RAIZ}/src" not in sys.path:
    sys.path.insert(0, f"{RAIZ}/src")

# Recarga limpia de los módulos (permite volver a ejecutar esta celda)
MODULOS = ["config", "interfaz", "validaciones", "archivos", "radicado",
           "autenticacion", "reportes", "registro", "consultas", "cambios", "main"]
for m in MODULOS:
    sys.modules.pop(m, None)

import config, interfaz, main

config.MOSTRAR_ASTERISCOS = False                          # en Colab la clave se lee con getpass
interfaz.limpiar_pantalla = lambda: print("\n" + "=" * 92)  # separador en lugar de limpiar

try:
    main.main()
except (KeyboardInterrupt, EOFError):
    print("\n\n  Programa finalizado por el usuario.")
```
